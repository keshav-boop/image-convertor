```python
import shutil
import os
import glob

paths = glob.glob("/usr/local/lib/python3.12/dist-packages/transformers*")

print("Removing:")
for p in paths:
    print(p)
    if os.path.isdir(p):
        shutil.rmtree(p, ignore_errors=True)
    else:
        try:
            os.remove(p)
        except:
            pass

print("DONE")
```
